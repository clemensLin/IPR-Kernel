# XXZ ground states: SVM balanced accuracies

Computes the balanced accuracies of Table IV for the $X$/$Z$ and $XY$/$Z$ kernels and writes them to `figures/tables/xxz_svm.json`, which `figures/make_figures.py` reads for Fig. 9(b).

Requires `data/processed/xxz/` from `process_data.ipynb`. The test fraction is 0.05 because the grid has 5000 points, as in the manuscript.

```python
from ipr_kernel.svm import run_iprkernel_svm_experiment
from ipr_kernel.tables import TABLES_DIR, svm_rows, write_table
```

```python
# Settings of the manuscript.
L_VALUES = list(range(6, 25, 2))
N_SPLITS = 100
COMMON = dict(dataset="heisenberg", L_values=L_VALUES, phase_boundary=1.0, test_frac=0.05, local_delta=0.25,
              rng_seed=12345, use_normalized_kernel=True, make_plot=False, print_report=True)
```

```python
results = {}
for key, basis in (("X_Z", ("IPR X", "IPR Z")), ("XY_Z", ("IPR XY", "IPR Z"))):
    print(f"--- {key}")
    _, _, results[key] = run_iprkernel_svm_experiment(basis=basis, n_splits=N_SPLITS, **COMMON)
```

```python
write_table(TABLES_DIR / "xxz_svm.json", {k: svm_rows(r["L"], r) for k, r in results.items()},
            f"Balanced accuracies of tab:svm_xxz ({N_SPLITS} splits, local delta = 0.25). "
            "Columns: L, global mean, global std, local mean, local std.")
```
